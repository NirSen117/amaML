```python
import pandas as pd

s1 = pd.read_csv(
    "dataset/train/train_source1.tsv",
    sep="\t"
)

s2 = pd.read_csv(
    "dataset/train/train_source2.tsv",
    sep="\t"
)

s3 = pd.read_csv(
    "dataset/train/train_source3.tsv",
    sep="\t"
)

gt = pd.read_csv(
    "dataset/train/train_ground_truth.tsv",
    sep="\t"
)

print("Source 1:", s1.shape)
print("Source 2:", s2.shape)
print("Source 3:", s3.shape)
print("Ground truth:", gt.shape)
```

```python
print("S1 columns:", s1.columns.tolist())
print("S2 columns:", s2.columns.tolist())
print("S3 columns:", s3.columns.tolist())
print("GT columns:", gt.columns.tolist())
```

```python
print("SOURCE 1")
print(s1.columns.tolist())
print(s1.dtypes)
display(s1.head(10))

print("\nSOURCE 2")
print(s2.columns.tolist())
print(s2.dtypes)
display(s2.head(10))

print("\nSOURCE 3")
print(s3.columns.tolist())
print(s3.dtypes)
display(s3.head(10))

print("\nGROUND TRUTH")
print(gt.columns.tolist())
print(gt.dtypes)
display(gt.head(10))
```

```python
gt["matched_list"] = gt["matched_entity_ids"].str.split(",")
gt["num_matches"] = gt["matched_list"].str.len()

print(gt["num_matches"].describe())
print(gt["num_matches"].value_counts().sort_index().head(30))
```

```python
# Split matched IDs, treating missing values as no matches
gt["matched_list"] = gt["matched_entity_ids"].fillna("").str.split(",")

gt["s2_matches"] = gt["matched_list"].apply(
    lambda x: [i for i in x if i.startswith("S2-") and i]
)

gt["s3_matches"] = gt["matched_list"].apply(
    lambda x: [i for i in x if i.startswith("S3-") and i]
)

gt["num_s2"] = gt["s2_matches"].str.len()
gt["num_s3"] = gt["s3_matches"].str.len()

print(gt[["num_s2", "num_s3"]].describe())
```

```python
for name, df in {
    "S1": s1,
    "S2": s2,
    "S3": s3
}.items():

    print(f"\n{name}")
    print(df["country"].value_counts().head(20))
```

```python
for name, df in {
    "S1": s1,
    "S2": s2,
    "S3": s3
}.items():

    print(f"\n{name}")
    print(df.isna().sum())
```

```python
for name, df in {"S1": s1, "S2": s2, "S3": s3}.items():
    print(f"\n{'='*50}")
    print(name)

    print("Duplicate entity_id:", df["entity_id"].duplicated().sum())

    print(
        "Duplicate business_name:",
        df["business_name"].duplicated().sum()
    )

    print(
        "Duplicate business_address:",
        df["business_address"].duplicated().sum()
    )

    print(
        "Duplicate name + address:",
        df[["business_name", "business_address"]]
        .duplicated()
        .sum()
    )
```
